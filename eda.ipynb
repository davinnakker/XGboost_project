```python
import polars as pl

df = pl.read_csv("https://raw.githubusercontent.com/byui-cse/cse450-course/master/data/housing.csv")
```

```python
df = df.filter(pl.col("bedrooms") <= 20)
```

```python
df.select(pl.max("bedrooms"))
df.write_csv("data/clean.csv")
```

```python
df.write_csv("data/raw.csv")
```

```python
df.describe()
["bedrooms", "bathrooms", "sqft_living", "sqft_lot", "floors", "sqft_above", "sqft_basement"]
```

```python
pl.Config.set_tbl_rows(70)
pl.Config.set_fmt_float("full")
(
    df
    .group_by("zipcode")
    .agg(
        mean_price = pl.col("price").mean(),
        median_price = pl.col("price").median(),
        max_price = pl.col("price").max(),
        built_mode = pl.col("yr_built").mode(),
        mean_sqft = pl.col("sqft_living15").mean(),
        median_bedrooms = pl.col("bedrooms").median(),
        median_bathrooms = pl.col("bathrooms").median(),
        median_floors = pl.col("floors").median(),
        median_condition = pl.col("condition").median(),
        median_grade = pl.col("grade").median(),
        avg_grade = pl.col("grade").mean(),
        num_of_houses = pl.len()
        )
    .sort("mean_price", descending=True)
)
```
