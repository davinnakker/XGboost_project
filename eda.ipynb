```python
import polars as pl
import numpy as np
from lets_plot import *
LetsPlot.setup_html()

df = pl.read_csv("data/raw.csv")
df = (
    df.select(["price", "long", "lat"])
    .filter(pl.col("long") < -121.6)
    .with_columns(np.log10(pl.col("price")))
)

(
    ggplot(data=df, mapping=aes(x='long', y='lat', color='price'))
    + geom_point(size=1.4)
    + scale_color_gradient(low="purple", high="yellow")
    + theme(
        panel_background=element_rect(fill="black"),
        plot_background=element_rect(fill="white"))
    + labs(title="Higher Priced Homes located new 'Eastside' of Seattle",
           subtitle="Home prices sky rocket as you get closer to Bellevue")
)
```
